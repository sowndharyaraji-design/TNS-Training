```python
#1. Reverse an Array

arr=[10,20,30,40,50]
rev=[]
for i in range(len(arr)-1,-1,-1):
    rev.append(arr[i])
print(rev)
```

```python

#2. Count Even, Odd and Zero
arr = [10,5,0,7,8,13,4]
even = 0
odd = 0
zero = 0

for i in arr:
    if i==0:
        zero +=1
    elif i%2==0:
        even+=1
    else:
        odd+=1
print("zero:" ,zero)
print("even:",even)
print("odd:",odd)
```

```python
#3. Sum of Positive and Negative Numbers
arr = [10,-5,20,-8,15,-2]
positiveSum=0
negativeSum=0
for i in arr:
    if i > 0:
        positiveSum = positiveSum + i
    elif i< 0:
        negativeSum = negativeSum + i

print(positiveSum)
print(negativeSum)
```

```python
#4. Remove Duplicate Elements
arr = [10,20,10,30,20,40,30]
unique=[]
for i in arr:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
#6. Rotate an Array
arr = [1, 2, 3, 4, 5]
k = 2
rotated = []
for i in range(len(arr)-k, len(arr)):
    rotated.append(arr[i])

for i in range(len(arr)-k):
    rotated.append(arr[i])

print(rotated)
```

```python
#7. Find the Most Frequent Element
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]
count = {}
most = arr[0]
max_count = 0
for i in arr:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1
for i in count:
    if count[i] > max_count:
        max_count = count[i]
        most = i 

print(count)
print("Most Frequent Element:", most)
print("Frequency:", max_count)
```

```python
#8. Maximum Subarray Sum
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = arr[0]
max_sum = arr[0]

for i in range(1, len(arr)):
    current_sum = max(arr[i], current_sum + arr[i])

    max_sum = max(max_sum, current_sum)

print("Maximum Subarray Sum:", max_sum)
```

```python
#9. Check Whether Two Arrays are Equal
arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

count1 = {}
count2 = {}

for i in arr1:
    if i in count1:
        count1[i] += 1
    else:
        count1[i] = 1

for i in arr2:
    if i in count2:
        count2[i] += 1
    else:
        count2[i] = 1

if count1 == count2:
    print("Arrays are Equal")
else:
    print("Arrays are Not Equal")
```
