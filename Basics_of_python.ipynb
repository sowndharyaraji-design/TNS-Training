```python
#Constructor

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


student1 = Student("Arun", 20)

student1.display()
```

```python
#OOPS - Inheritance

class Animal:

    def eat(self):
        print("Animal eats")


class Dog(Animal):

    def bark(self):
        print("Dog barks")


dog1 = Dog()

dog1.eat()
dog1.bark()
```

```python
#OOPS - Encapsulation

class Student:

    def __init__(self, name, mark):
        self.name = name
        self.__mark = mark

    def display(self):
        print("Name:", self.name)
        print("Mark:", self.__mark)


student = Student("Arun", 85)

student.display()
```

```python
#OOPS - Polymorphism

class Dog:

    def sound(self):
        print("Dog says Woof")


class Cat:

    def sound(self):
        print("Cat says Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
#OOPS - Absraction

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts with a key")


car = Car()
car.start()
```
