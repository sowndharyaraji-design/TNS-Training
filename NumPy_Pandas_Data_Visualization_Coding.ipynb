```python
#2. NumPy – Temperature Analysis
import numpy as np

temp = np.array([28, 32, 29, 35, 31, 27, 33])

print("Temperatures:", temp)
print("Average:", np.mean(temp))
print("Highest:", np.max(temp))
print("Lowest:", np.min(temp))
print("Above 30°C:", temp[temp > 30])

updated = temp + 2
print("Updated temperatures:", updated)
```

```python
#3. Pandas – Student Performance DataFrame


import pandas as pd

data = {
    "Name": ["Anu", "Bala", "Cathy", "David",
             "Esha", "Farhan", "Gita", "Hari"],
    "Department": ["CSE", "IT", "CSE", "ECE",
                   "IT", "CSE", "ECE", "IT"],
    "Marks": [85, 72, 90, 65, 78, 95, 56, 81],
    "Attendance": [90, 75, 85, 70, 95, 88, 78, 82]
}

df = pd.DataFrame(data)

print("First 5 students:")
print(df.head())

print("\nAverage marks:", df["Marks"].mean())

print("\nStudents scoring above 75:")
print(df[df["Marks"] > 75])

print("\nAttendance below 80%:")
print(df[df["Attendance"] < 80])

print("\nStudents sorted by marks:")
print(df.sort_values(by="Marks"))
```

```python
#4. Pandas – Product Sales Analysis

import pandas as pd

data = {
    "Product Name": ["Laptop", "Mouse", "Keyboard",
                     "Monitor", "Headphones"],
    "Category": ["Electronics", "Accessories",
                 "Accessories", "Electronics", "Audio"],
    "Price": [50000, 500, 1500, 12000, 2000],
    "Quantity Sold": [10, 80, 60, 20, 55]
}

df = pd.DataFrame(data)

df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("Product Sales:")
print(df)

print("\nHighest sales product:")
print(df.loc[df["Total Sales"].idxmax()])

print("\nAverage product price:", df["Price"].mean())

print("\nQuantity sold above 50:")
print(df[df["Quantity Sold"] > 50])

print("\nProducts sorted by total sales:")
print(df.sort_values(by="Total Sales", ascending=False))
```

```python
#5. Data Visualization – Monthly Sales

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10000, 15000, 12000, 18000, 22000, 20000]

# Line chart
plt.plot(months, sales, marker="o", label="Monthly Sales")
plt.title("Monthly Sales Line Chart")
plt.xlabel("Months")
plt.ylabel("Sales (Rs.)")
plt.legend()
plt.grid(True)
plt.show()

# Bar chart
plt.bar(months, sales, label="Monthly Sales")
plt.title("Monthly Sales Bar Chart")
plt.xlabel("Months")
plt.ylabel("Sales (Rs.)")
plt.legend()
plt.show()
```

```python
#6. Data Visualization – Student Marks

import matplotlib.pyplot as plt

students = ["S1", "S2", "S3", "S4", "S5",
            "S6", "S7", "S8", "S9", "S10"]

marks = [95, 82, 75, 68, 55, 45, 35, 88, 60, 25]

# Bar chart
plt.bar(students, marks, label="Student Marks")
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()

# Performance categories
excellent = sum(80 <= m <= 100 for m in marks)
good = sum(60 <= m < 80 for m in marks)
average = sum(40 <= m < 60 for m in marks)
needs_improvement = sum(m < 40 for m in marks)

categories = ["Excellent", "Good", "Average",
              "Needs Improvement"]

counts = [excellent, good, average, needs_improvement]

# Pie chart
plt.pie(counts, labels=categories, autopct="%1.1f%%")
plt.title("Student Performance Distribution")
plt.show()
```
